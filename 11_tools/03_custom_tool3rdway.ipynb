## 3rd method using `BaseTool`

```python
# No additional import is needed; use type[BaseModel] as defined below.

from langchain.tools import BaseTool
from pydantic import BaseModel, Field
```

```python
class MultiplyArgs(BaseModel):
    a: int = Field(..., description="The first number to multiply")
    b: int = Field(..., description="The second number to multiply")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiplies two numbers together."
    args_schema: type[BaseModel] = MultiplyArgs

    def _run(self, a: int, b: int) -> int:
        return a * b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({"a": 5, "b": 10})
print(result)
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args_schema.model_json_schema())
```
