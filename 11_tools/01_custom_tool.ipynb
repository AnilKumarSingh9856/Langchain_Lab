# Custom Tools

```python
from langchain_core.tools import tool
```

## Step 1: Create a function

```python
def multiply1(a: float, b: float) -> float:
    """ Multiplies two numbers and returns the result."""
    return a * b
```

## Step 2: add type hints to the function

```python
def multiply2(a: float, b: float) -> float:
    """ Multiplies two numbers and returns the result."""
    return a * b
```

## Step 3: add tool decorator to the function

```python
@tool
def multiply3(a: float, b: float) -> float:
    """ Multiplies two numbers and returns the result."""
    return a * b
```

```python
result = multiply3.invoke({'a': 3, 'b': 4})
```

```python
print(f"The result of multiplying 3 and 4 is: {result}")
```

```python
print(multiply3.name)
print(multiply3.description)
print(multiply3.args)
```

```python
print(multiply3.args_schema.model_json_schema())
```

```python
schema = multiply3.args_schema.model_json_schema()
for prop, details in schema.items():
    print(f"Property: {prop}")
    print(f"Details: {details}")
    print()
```
