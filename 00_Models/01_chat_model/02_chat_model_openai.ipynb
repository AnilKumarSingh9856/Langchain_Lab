```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
ChatOpenAI.api_key = os.getenv("OPENROUTER_API_KEY")
```

```python
model = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=ChatOpenAI.api_key,
    base_url="https://openrouter.ai/api/v1",
    max_tokens=512,

)
```

```python
result = model.invoke("Hello, how are you?")
```

```python
print(result.content)
```

```python
result_token_usage = result.response_metadata
print(result_token_usage)  # Print token usage information
```

```python
token_usage               = result_token_usage.get("token_usage", {})
completion_tokens_details = result_token_usage.get("completion_tokens_details", {})
prompt_tokens_details     = result_token_usage.get("prompt_tokens_details", {})
```

```python
print(token_usage)
print(completion_tokens_details)
print(prompt_tokens_details)
```

```python
completion_tokens = result.response_metadata.get("token_usage", {}).get("completion_tokens", 0)
prompt_tokens = result.response_metadata.get("token_usage", {}).get("prompt_tokens", 0)
total_tokens = result.response_metadata.get("token_usage", {}).get("total_tokens", 0)
print(f"Completion Tokens : {completion_tokens}")
print(f"Prompt Tokens     : {prompt_tokens}")
print(f"Total Tokens      : {total_tokens}")
```
