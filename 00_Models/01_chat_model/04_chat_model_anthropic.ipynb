```python
import os
from dotenv import load_dotenv
from langchain_anthropic import ChatAnthropic
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
api_key = os.getenv("OPENROUTER_API_KEY")
if not api_key:
    raise RuntimeError("Set OPENROUTER_API_KEY in your .env file.")

model = ChatAnthropic(
    model="anthropic/claude-haiku-4.5",
    api_key=api_key,
    base_url="https://openrouter.ai/api",
    max_tokens=512,
)
```

```python
result = model.invoke("Hi, how are u?")
```

```python
print(result.content)
```

```python
print(result.response_metadata.get("usage").get("input_tokens"))  # Print the usage information
print(result.response_metadata.get("usage").get("output_tokens"))  # Print the usage information
```
