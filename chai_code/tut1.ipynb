```python
import os

from openai import OpenAI
```

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
model_name = os.getenv("GEMINI_MODEL", "gemini-3.5-flash-lite")
```

```python
client = OpenAI(
    api_key=os.getenv("GEMINI_API_KEY", ""),
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
)
```

```python
response = client.chat.completions.create(
    model = model_name,
    messages = [
        {
            "role" : "user",
            "content": "hey , how are u"
        }
    ]
)

print(response.choices[0].message.content)
```

```python
print(list(response.usage.model_dump().items())[:3])
```

```python
SYSTEM_PROMPT = """
Rule no 1. You are coding expert in all programming language. U only respond when q is related to coding otherwise u response sorry, i have no idea like professional message.

Rule no 2. You strictly response in json format only.
exp  : {{"is_coding" : false or true, "data" : null or content(str)}}

examples :
q1. what is your name ?
 answer : {{"is_coding" : false, "data" : null}}

q2. what is the sum of two number a and b
answer : {{"is_coding" : true, "data" : def sums(a, b): return a+b}}
"""
```

```python
response1 = client.chat.completions.create(
    model = model_name,
    messages = [
        {
            "role" : "user",
            "content": "hey, Could u please tell me how to find second highest number without sorting"
        },
        {
            "role" : "system",
            "content": SYSTEM_PROMPT
        }
    ]
)
```

```python
print(list(response1.usage.model_dump().items())[:3])
```

```python
print(response1.choices[0].message.content)
```
