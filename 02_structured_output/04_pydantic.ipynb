```python
class Person(BaseModel):
    name : str
    age : int
    gender : Optional[str] = None
```

```python
new_person = {
    "name": "Anil",
    "age":  20,
}
```

```python
print(new_person)
```

```python
new_person_1 : Person = {
    "name" : 20,
    "age" : "anil"

}
```

```python
print(new_person_1)
```

```python
person = Person(**new_person)
print(person)
```
