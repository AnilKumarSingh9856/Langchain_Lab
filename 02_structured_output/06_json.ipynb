A JSON Schema is a JSON document that describes what a valid JSON object should look like. It specifies the data types, required fields, allowed values, and so on.

Simple example
Suppose a person must have a name and an age:

```python
{
  "type": "object",
  "properties": {
    "name": { "type": "string" },
    "age": { "type": "integer" }
  },
  "required": ["name", "age"]
}
```

```python
{
  "name": "Anil",
  "age": 20
}
```

```python
schema = {
    "type": "object",
    "properties": {"name": {"type": "string"}, "age": {"type": "integer"}},
    "required": ["name", "age"],
}
```

```python
print(schema)
```
