```python
from typing import  TypedDict
```

```python
class Person(TypedDict):
    name : str
    age : int
```

```python
new_person : Person = {
    'name' : 'nitish',
    'age' :  20
}
```

```python
print(new_person)
```

Typedict only give hint or suggestion that what type of value u need to assign.
It deos not validate

```python
second_Person: Person = {
    'name' : 20,
    'age' : 'nitish'
}
```

```python
print(second_Person)
```

From above example we can  see that it does not throwing error, even after changing its data type of its value. It only shows suggestion when u hover on it.
