```python
from langchain_experimental.text_splitter import SemanticChunker
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv
import os
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
text_splitter = SemanticChunker(
    OpenAIEmbeddings(
        model="text-embedding-3-large",
        dimensions=30,
        api_key=os.getenv("OPENROUTER_API_KEY"),
        base_url="https://openrouter.ai/api/v1",
    ),
    breakpoint_threshold_type="standard_deviation",
    breakpoint_threshold_amount=0.5,
)
```

```python
text = """
Electric vehicles are becoming increasingly popular as an alternative to traditional petrol-powered cars. They use rechargeable batteries and electric motors to reduce fuel consumption and tailpipe emissions. Improvements in battery technology and the expansion of charging infrastructure are making electric cars more practical for everyday transportation.

Solar energy is gaining popularity as a renewable source of electricity. Solar panels convert sunlight into electrical power, helping households and businesses reduce their dependence on conventional energy sources. Advances in photovoltaic technology and improvements in energy storage are making solar power more affordable and accessible.

The human brain contains billions of neurons that communicate through electrical and chemical signals. These connections enable people to learn new skills, store memories, process emotions, and make decisions. Neuroscientists study brain activity to understand cognition and investigate neurological disorders such as Alzheimer's disease.
"""
```

```python
splits = text_splitter.split_text(text)
```

```python
print("Splits:")
for i, split in enumerate(splits):
    print(f"Split {i + 1}: {split}\n")
```
