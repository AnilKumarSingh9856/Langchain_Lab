````python
from langchain_text_splitters import RecursiveCharacterTextSplitter
````

````python
text = """
Python is a high-level, interpreted programming language.

It is widely used in web development, data science, artificial intelligence,
automation, and backend engineering.

Python supports procedural, object-oriented, and functional programming.

Variables and Data Types
========================

Variables store values used throughout a program. Python is dynamically typed.

Common data types include integers, floats, strings, booleans, lists,
tuples, dictionaries, and sets.

Functions
=========

Functions organize reusable logic into named blocks of code.

Example:

def calculate_total(price, quantity):
    return price * quantity

total = calculate_total(100, 3)
print(total)

Object-Oriented Programming
===========================

Object-oriented programming organizes code around classes and objects.

Example:

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_details(self):
        return f"{self.name}: {self.price}"

product = Product("Keyboard", 50)
print(product.get_details())

Exception Handling
==================

Exceptions occur when a program encounters an error during execution.
Python provides try, except, else, and finally blocks to handle errors.

File Handling
=============

Python provides built-in functions for reading and writing files.
The with statement ensures that files are properly closed after use.

Backend Development
===================

FastAPI is a Python framework for building APIs. It supports asynchronous
request handlers, dependency injection, and request validation.

Example:

from fastapi import FastAPI

app = FastAPI()

@app.get("/products/{product_id}")
async def get_product(product_id: int):
    return {"product_id": product_id}

Best Practices
==============

1. Use descriptive names.
2. Keep functions focused.
3. Handle errors explicitly.
4. Write automated tests.
5. Separate business logic from API routes.

Conclusion
==========

Python is suitable for projects ranging from simple scripts to distributed
backend systems. Understanding its core features provides a foundation
for learning frameworks, databases, testing, and scalable architecture.
"""
````

````python
from langchain_text_splitters import Language

loader = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON,
    chunk_size=500,
    chunk_overlap=50
)
````

````python
splits = loader.split_text(text)
````

````python
print(f"Number of splits: {splits[0]}")
````

````python
markdown_text = """
# Python Programming Guide

Python is a high-level, interpreted programming language used in web
development, data science, artificial intelligence, and automation.

## Variables and Data Types

Python is dynamically typed.

Common data types include:
- `int`: Integer numbers
- `float`: Decimal numbers
- `str`: Text
- `bool`: Boolean values
- `list`: Mutable collections
- `tuple`: Immutable collections
- `dict`: Key-value mappings
- `set`: Unique elements

## Functions

Functions organize reusable logic.

```python
def calculate_total(price, quantity):
    return price * quantity

print(calculate_total(100, 3))

"""
````

````python
markdown_loader = RecursiveCharacterTextSplitter.from_language(
    language=Language.MARKDOWN,
    chunk_size=500,
    chunk_overlap=50
)
````

````python
splits = markdown_loader.split_text(markdown_text)
````

````python
print(f"Number of splits: {len(splits)}")
````

````python
print(splits[0])
````
