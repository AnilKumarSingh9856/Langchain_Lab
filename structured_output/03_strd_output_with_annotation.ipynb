```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os
from typing import TypedDict, List, Literal, Optional, Annotated
```

```python
model = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
    max_tokens=512,
)
```

```python
class Review(TypedDict):
    key_themes: Annotated[
        list[str],
        "Identify all major topics, features, issues, and points discussed in the review. Represent each theme as a concise, distinct phrase. Do not include information that is not supported by the review.",
    ]
    summary: Annotated[
        str,
        "Provide a concise, objective summary of the review, capturing the reviewer's main opinion, key observations, and overall experience without adding assumptions.",
    ]
    sentiment: Annotated[
        Literal["+ive", "-ive", "0"],
        "Classify the overall sentiment as positive (+ive), negative (-ive), or neutral (0).",
    ]
    pros: Annotated[
        Optional[list[str]],
        "Extract the specific advantages, positive features, benefits, and satisfactory experiences explicitly mentioned by the reviewer. Return None if no pros are mentioned.",
    ]
    cons: Annotated[
        Optional[list[str]],
        "Extract the specific disadvantages, problems, defects, limitations, complaints, and unsatisfactory experiences explicitly mentioned by the reviewer. Return None if no cons are mentioned.",
    ]
    name: Annotated[
        Optional[str],
        "Extract the reviewer's name if explicitly present; otherwise return None. Never generate a title.",
    ]
```

```python
review_text = """
I purchased these wireless headphones two weeks ago. The sound quality
is excellent, and the battery lasts almost 30 hours. They are comfortable
to wear for long periods. However, the microphone quality is poor during
calls, and the plastic build feels cheap. Overall, I am satisfied with
the purchase. — Rahul Sharma
"""
```

```python
structured_output = model.with_structured_output(Review)
```

```python
result = structured_output.invoke(review_text)
```

```python
for key, value in result.items():
    print(f'{key} ----> {value}')
```
