```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
```

```python
model = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
    max_tokens=512,
)
```

```python
parser =StrOutputParser()
```

```python
template1 = PromptTemplate(
    template="Give me detailed report about this topic: {topic}.",
    input_variables=["topic"],
    validate_template=True,
)
```

```python
template2 = PromptTemplate(
    template="Give me five important key points from this detailed report: {detailed_report}.",
    input_variables=["detailed_report"],
    validate_template=True,
)
```

```python
chain1 = template1 | model | parser | template2 | model | parser
```

```python
result1 = chain1.invoke({"topic": "Son Goku"})
```

```python
print(result1)
```

```python
chain2= template1 | model |  template2 | model | parser
```

```python
result2 = chain2.invoke({"topic": "Attention Is All You Need"})
```

```python
print(result2)
```

```python
chain1.get_graph().print_ascii()
```

```python
chain2.get_graph().print_ascii()
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate

parser = StrOutputParser()

report_chain = template1 | model | parser

final_chain = {"detailed_report": report_chain} | template2 | model | parser

result = final_chain.invoke({"topic": "Son Goku"})

print(result)
```
