```python
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_anthropic import ChatAnthropic
from langchain_core.prompts import PromptTemplate
import os
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import  RunnableParallel
```

```python
load_dotenv()
```

```python
openai_model1 = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
    max_tokens=2048,
)
```

```python
prompt1 = PromptTemplate(
    template = 'Generate short and simple notes from the following text: {text}',
    input_variables = ['text'],
    validate_template = True
)
```

```python
google_model2 = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    max_output_tokens=2048,
)
```

```python
prompt2 = PromptTemplate(
    template = 'Generate five mcqs based on the following text: {text}',
    input_variables = ['text'],
    validate_template = True
)
```

```python
anthropic_model3 = ChatAnthropic(
    model="anthropic/claude-haiku-4.5",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api",
    max_tokens=2048,
)
```

```python
merge_prompt = PromptTemplate(
    template="""Create the final response using the following content.

    NOTES:
    {notes}

    QUIZ QUESTIONS:
    {quiz}

    Your response must contain:
    1. A notes section
    2. A quiz questions section
    3. All five quiz questions exactly as provided
    4. In the end thier solutions.
    """,
    input_variables=["notes", "quiz"],
)
```

```python
parser = StrOutputParser()
```

```python
parallel_chain = RunnableParallel({
    'notes': prompt1 | openai_model1 | parser,
    'quiz': prompt2 | google_model2 | parser,
})
```

```python
merged_chain = merge_prompt | anthropic_model3 | parser
```

```python
final_chain = parallel_chain | merged_chain
```

```python
result = final_chain.invoke({'text' : langchain_description})
```
