```python
from langchain_core.prompts import  ChatPromptTemplate
```

```python
template = ChatPromptTemplate([
        ('system' , 'you are a good scholor in {domain}'),
        ('human', 'tell me about {topic}')
    ]
)
```

```python
prompt = template.invoke({'domain': 'aerospace engineer', 'topic':'black hole'})
```
