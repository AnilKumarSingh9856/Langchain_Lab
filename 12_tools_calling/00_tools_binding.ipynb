```python
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI
import os
import requests
```

```python
# tool create

@tool
def multiply(a: int, b: int) -> int:
    """Multiplies two numbers together and returns their product."""
    return a * b
```

```python
print(multiply.invoke({'a': 5, 'b': 6}))
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args_schema)
```

### tool binding


```python
llm  = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.environ.get("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
)
```

```python
llm_with_tools = llm.bind_tools([multiply])
```

### tool calling

```python
llm_with_tools.invoke([HumanMessage(content="Hello, how are you?")])
```

```python
query = HumanMessage(content="What is 5 multiplied by 6?")
```

```python
message = [query]
```

```python
message
```

```python
result = llm_with_tools.invoke(message)
```

```python
result
```

```python
message.append(result)
```

### tool execution

```python
result = llm_with_tools.invoke([HumanMessage(content="What is 5 multiplied by 6?")])
print(result.tool_calls[0])  # This will print the result of the multiplication
```

```python
multiply.invoke(result.tool_calls[0]['args'])  # This will execute the multiply tool with the extracted arguments
```

```python
result = multiply.invoke(result.tool_calls[0])
result
```
