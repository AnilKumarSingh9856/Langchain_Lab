```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
```

```python
url = "https://www.amazon.in/iPhone-16-128-GB-Control/dp/B0DGJHBX5Y/?_encoding=UTF8&pd_rd_w=ASc7i&content-id=amzn1.sym.29dc001b-6f7b-4178-81f2-dbcb2f97d6bb&pf_rd_p=29dc001b-6f7b-4178-81f2-dbcb2f97d6bb&pf_rd_r=TG0XSFDZMC0T9HV3X0H5&pd_rd_wg=vOdr4&pd_rd_r=333d005d-e89a-45d1-b127-c3e3911383ef"
loader = WebBaseLoader(url)
```

```python
docs = loader.load()
```

```python
print(len(docs))
```

```python
print(docs[0].page_content)
```

```python
openai_model1 = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
    max_tokens=2048,
)
```

```python
prompt = PromptTemplate(
    template="Write a answer of the following questins: {questions} and the answer should be in the context of the following document: {document}",
    input_variables=["questions", "document"],
)
```

```python
parser = StrOutputParser()
```

```python
chain = prompt | openai_model1 | parser
```

```python
result = chain.invoke(
    {
        "questions": "What is the price of the product?",
        "document": docs[0].page_content,
    }
)
```

```python
print(result)
```
