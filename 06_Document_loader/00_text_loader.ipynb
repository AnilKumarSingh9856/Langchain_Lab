# Loading a text document

`TextLoader` is provided by the `langchain-community` package.

```python
from langchain_community.document_loaders import TextLoader
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
```

```python
loader = TextLoader("../01_prompts/09a_chat_history.txt", encoding="utf-8")
documents = loader.load()
```

```python
print(f"Loaded {len(documents)} document(s)")
print(documents[0].page_content)
```

```python
openai_model1 = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
    max_tokens=2048,
)
```

```python
prompt = PromptTemplate(
    template="You are a helpful assistant. Please write few lines about following person:\n\n{name}",
    input_variables=["name"],
)
```

```python
parser = StrOutputParser()
```

```python
chain = prompt | openai_model1 | parser
```

```python
result = chain.invoke({"name": documents[0].page_content})
```

```python
print(result)
```
