```python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader
```

```python
loader = DirectoryLoader(
    "../books", glob="**/*.pdf", loader_cls=PyPDFLoader, show_progress=True
)
```

```python
docs = loader.load()
```

```python
docs1 = loader.lazy_load()
```

```python
for doc in docs:
    print(doc.metadata)
```

```python
for doc in docs1:
    print(doc.metadata)
```
