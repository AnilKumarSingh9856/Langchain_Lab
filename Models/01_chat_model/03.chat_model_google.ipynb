```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
```

```python
model = ChatGoogleGenerativeAI(
    model = "gemini-3.5-flash-lite",
    max_output_tokens = 512,
)
```

```python
result = model.invoke("Hello, how are u?")
```

```python
print(result.content[0].get("text"))
```

```python
print(result.usage_metadata.get("input_tokens"))  # Print the usage information
print(result.usage_metadata.get("output_tokens"))  # Print the usage information
print(result.usage_metadata.get("total_tokens"))  # Print the usage information
```
