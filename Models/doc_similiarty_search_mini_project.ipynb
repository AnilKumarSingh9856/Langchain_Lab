```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv
import os
from sklearn.metrics.pairwise import cosine_similarity
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
embeddings = OpenAIEmbeddings(
    model="text-embedding-3-large",
    dimensions=303,
    api_key=os.getenv('OPENROUTER_API_KEY'),
    base_url="https://openrouter.ai/api/v1"
)
```

```python
document = [
    "John is a software engineer who loves to code in Python and JavaScript.",
    "Jane is a data scientist who loves to analyze data in R and Python.",
    "Bob is a product manager who loves to lead teams and drive innovation.",
    "Alice is a UX designer who loves to create intuitive user experiences."
]

query = "Who is John and what does he love?"
```

```python
document_vectors = embeddings.embed_documents(document)
  # Extract the first element of each embedding vector
```

```python
document_vectors
```

```python
query_vector = embeddings.embed_query(query)
```

```python
print(query_vector)  # Print the embedding vector for the query
```

```python
similarities = cosine_similarity([query_vector], document_vectors)
```

```python
print(similarities)
```

```python
print(list(enumerate(similarities[0])))  # Print documents with their similarity scores
```

```python
print(sorted(list(enumerate(similarities[0])), key=lambda x: x[1], reverse=True))  # Print documents with their similarity scores
```

```python
print(sorted(list(enumerate(similarities[0])), key=lambda x: x[1], reverse=True)[0])  # Print the most similar document with its similarity score
```

```python
index, score = sorted(list(enumerate(similarities[0])), key=lambda x: x[1], reverse=True)[0]
```

```python
print(query)  # Print the query
print(document[index])  # Print the most similar document
print(score)  # Print the similarity score
```
