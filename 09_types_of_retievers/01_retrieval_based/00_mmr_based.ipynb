```python
from langchain_core.documents import Document
```

```python
# sample dcuments
docs = [
    Document(page_content="Langchain makes it easy to build applications with large language models.", metadata={"category": "framework"}),
    Document(page_content="Langchain is used for building applications that can understand and generate natural language.", metadata={"category": "framework"}),
    Document(page_content="Chroma is a vector database that can be used to store and retrieve embeddings.", metadata={"category": "vector database"}),
    Document(page_content="OpenAI Embeddings is a library for generating embeddings.", metadata={"category": "embedding library"}),
    Document(page_content="OpenAI provides a set of APIs that can be used to generate embeddings for text, images, and other data types.", metadata={"category": "API"}),
    Document(page_content="MMR is a retrieval method that selects documents based on their relevance and diversity.", metadata={"category": "retrieval method"}),
    Document(page_content="The cosine similarity is a measure of similarity between two non-zero vectors.", metadata={"category": "similarity measure"}),
    Document(page_content="Langchain suports Chroma, FAISS, and other vector databases for storing and retrieving embeddings.", metadata={"category": "framework"}),
]
```

```python
import os

from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings
```

```python
embeddings_model = OpenAIEmbeddings(
    model="text-embedding-3-large",
    dimensions=30,
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
)
```

```python
# create a FAISS vector store from the documents and embeddings model
vector_store = FAISS.from_documents(
    documents=docs,
    embedding=embeddings_model

)
```

```python
# Enable MMR retrieval
retriever = vector_store.as_retriever(search_type="mmr", search_kwargs={"k": 3, "lambda": 0.5})
```

```python
query = "What is Langchain?"
results = retriever.invoke(query)
```

```python
for i, result in enumerate(results):
    print(f"Result {i+1}:")
    print(f"Content: {result.page_content}")
    print(f"Metadata: {result.metadata}")
    print()
```
