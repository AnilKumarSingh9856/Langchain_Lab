```python
import os

from langchain_chroma import Chroma
from langchain_core.documents import Document
from langchain_openai import OpenAIEmbeddings
```

```python
# Step 1 : Your source documents
documents = [
    Document(page_content="Langchain is a powerful framework for building language model applications", metadata={"category": "framework"}),
    Document(page_content="Chroma is a vector database that can be used to store and retrieve embeddings", metadata={"category": "vector database"}),
    Document(page_content="OpenAI Embeddings is a library for generating embeddings", metadata={"category": "embedding library"}),
    Document(page_content="OpenAI provides a set of APIs that can be used to generate embeddings for text, images, and other data types", metadata={"category": "API"}),
]
```

```python
# step 2 Initialize the embeddings and vector store
embeddings_model = OpenAIEmbeddings(
    model="text-embedding-3-large",
    dimensions=30,
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
)

vector_store = Chroma.from_documents(
    documents = documents,
    embedding =embeddings_model,
    persist_directory="./my_collection",
)
```

```python
# step 3 : Convert vector store to retriever
retriever = vector_store.as_retriever(search_type="similarity", search_kwargs={"k": 2})
```

```python
# step 4 : Use the retriever to retrieve relevant documents based on a query
query = "What is Langchain?"
results = retriever.invoke(query)
```

```python
for i, result in enumerate(results):
    print(f"Result {i+1}:")
    print(f"Content: {result.page_content}")
    print(f"Metadata: {result.metadata}")
    print()
```
