```python
from langchain_community.retrievers import WikipediaRetriever
```

```python
import wikipedia.wikipedia as wikipedia_api

wikipedia_api.API_URL = "https://en.wikipedia.org/w/api.php"
wikipedia_api.USER_AGENT = "LangchainTut/1.0 (educational project)"
```

```python
# Initialize the WikipediaRetriever ( optional: set language and top_k )
retriever = WikipediaRetriever(language="en", top_k_results=3,)
```

```python
# Define a query to search for
query = "India and Pakistan relations based on perspective of Kashmir"
```

```python
# Get the relevant documents from Wikipedia
docs = retriever.invoke(query)
```

```python
# Print the retrieved documents
print("Number of documents:", len(docs))

if not docs:
    print("No documents found.")
else:
    for i, doc in enumerate(docs, start=1):
        print(f"\nDocument {i}")
        print("Title:", doc.metadata.get("title", "N/A"))
        print("Content:")
        print(doc.page_content[:1000])
        print("-" * 80)
```
