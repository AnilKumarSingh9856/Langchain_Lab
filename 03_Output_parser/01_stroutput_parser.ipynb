```python
from langchain_huggingface import ChatHuggingFace, HuggingFacePipeline
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = HuggingFacePipeline.from_model_id(
    model_id="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    task="text-generation",
    pipeline_kwargs={
        "max_new_tokens": 2048,
        "temperature": 0.7,
        "top_p": 0.7,
        "repetition_penalty": 1.1,
    },
)
```

```python
model = ChatHuggingFace(
    llm=llm,
)
```

```python
template1 = PromptTemplate(
    template="You are a helpful assistant. Write a detailed description of the following topic:\n\n{topic}",
    input_variables=["topic"],
)
```

```python
template2 = PromptTemplate(
    template="You are a helpful assistant. Write a short summary of the following text:\n\n{text}",
    input_variables=["text"],
)
```

```python
chain = template1 | model | StrOutputParser() | template2 | model | StrOutputParser()
```

```python
result = chain.invoke("Black holes in space")
```
