```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
import os
from langchain_core.output_parsers import StrOutputParser
```

```python
model = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
    max_tokens=1024,
)
```

```python
template1 = PromptTemplate(
    template="You are a helpful assistant. Write a detailed description of the following topic:\n\n{topic}",
    input_variables=["topic"],
)
```

```python
template2 = PromptTemplate(
    template="You are a helpful assistant. Write a short summary of the following text:\n\n{text}",
    input_variables=["text"],
)
```

```python
chain = template1 | model | StrOutputParser() | template2 | model | StrOutputParser()
```

```python
result = chain.invoke("Naruto")
```
