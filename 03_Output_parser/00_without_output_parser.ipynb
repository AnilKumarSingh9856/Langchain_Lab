```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
import os
```

```python
model = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
    max_tokens=512,
)
```

```python
template1 = PromptTemplate(
    template="You are a helpful assistant that translates {user_input} from {input_language} to {output_language}.",
    input_variables=["input_language", "output_language"]
)
```

```python
prompt1 = template1.format(
    user_input = "Hello, how are you?",
    input_language="English",
    output_language="French",
)
```

```python
result = model.invoke(prompt1)
```

```python
print(result.content)
```

```python
template2 = PromptTemplate(
    template = 'Summarize the following contents {text}',
    input_variables = ['text']
)
```

```python
prompt2 = template2.invoke(result.content)
```

```python
result = model.invoke(prompt2)
```

```python
print(result.content)
```
