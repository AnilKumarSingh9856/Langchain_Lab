```python
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_anthropic import ChatAnthropic
from langchain_core.prompts import PromptTemplate
import os
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel, RunnableSequence
```

```python
openai_model1 = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
    max_tokens=2048,
)
```

```python
prompt1 = PromptTemplate(
    template = 'Write  a tweet about this topic: {topic}',
    input_variables = ['topic'],
)
```

```python
parser = StrOutputParser()
```

```python
google_model2 = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    max_output_tokens=2048,
)
```

```python
prompt2 = PromptTemplate(
    template = 'Write a linkdin post about this topic: {topic}',
    input_variables = ['topic'],
)
```

```python
parallel_chain = RunnableParallel({
    'tweet' : RunnableSequence(prompt1, openai_model1, parser),
    'linkdin' : RunnableSequence(prompt2, google_model2, parser)
})
```

```python
result = parallel_chain.invoke({'topic' : 'time management'})
```

```python
for key, value in result.items():
    print(f'{key} ----> {value}')
```
