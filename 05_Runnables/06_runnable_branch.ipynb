```python
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_anthropic import ChatAnthropic
from langchain_core.prompts import PromptTemplate
import os
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import (
    RunnableSequence,
    RunnablePassthrough,
    RunnableBranch,
)
```

```python
openai_model1 = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
    max_tokens=2048,
)
```

```python
prompt1 = PromptTemplate(
    template = "Write details report about topic: {topic}",
    input_variables = ["topic"],
)
```

```python
prompt2 = PromptTemplate(
    template = "Write a short summary of this report:\n{report}",
    input_variables = ["report"],
)
```

```python
parser = StrOutputParser()
```

```python
detail_report = RunnableSequence(prompt1, openai_model1, parser)
```

```python
branch_chain = detail_report | RunnableBranch(
    (lambda report: len(report.split()) > 500, RunnableSequence(prompt2, openai_model1, parser)),
    RunnablePassthrough()
)
```

```python
result = branch_chain.invoke({'topic': 'Russia Ukraine war'})
```
