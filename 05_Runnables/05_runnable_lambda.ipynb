```python
import os

from dotenv import load_dotenv
from langchain_anthropic import ChatAnthropic
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import (
    RunnableLambda,
    RunnableParallel,
    RunnablePassthrough,
    RunnableSequence,
)
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_openai import ChatOpenAI
```

```python
def text_count(text: str) -> int:
    """Count the number of words in a string."""
    return len(text.split())
```

```python
runnable_lambda = RunnableLambda(text_count)
```

```python
print(runnable_lambda.invoke("This is a test string to count the number of words."))  # Output: 11
```

```python
openai_model1 = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
    max_tokens=2048,
)
```

```python
prompt1 = PromptTemplate(
    template="Write  a joke about this topic: {topic}",
    input_variables=["topic"],
)
```

```python
parser = StrOutputParser()
```

```python
joke_gen_chain = RunnableSequence(prompt1, openai_model1, parser)
```

```python
parallel_chain = RunnableParallel({
    'joke' : RunnablePassthrough(),
    'word_count' : RunnableLambda(text_count)
})
```

```python
combined_chain = RunnableSequence(joke_gen_chain, parallel_chain)
```

```python
result = combined_chain.invoke("Liberty and justice for all")
```

```python
print("""The joke is: {joke}
The number of words in the joke is: {word_count}""".format(**result))
```
