````python
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_anthropic import ChatAnthropic
from langchain_core.prompts import PromptTemplate
import os
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import (
    RunnableSequence,
    RunnablePassthrough,
    RunnableBranch,
)
````

````python
openai_model1 = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url="https://openrouter.ai/api/v1",
    max_tokens=2048,
)
````

````python
prompt1 = PromptTemplate(
    template="Write details report about topic: {topic}",
    input_variables=["topic"],
)
````

````python
parser = StrOutputParser()
````

````python
detail_report = RunnableSequence(prompt1, openai_model1, parser)
````

that is langchain expression language. It is a way to chain together different components in LangChain using a simple syntax. In this case, we are chaining a prompt, an OpenAI model, and a parser together to create a detailed report. The `|` operator is used to indicate the flow of data from one component to the next.
Example:
```python
detail_report1 = prompt1 | openai_model1 | parser
```
